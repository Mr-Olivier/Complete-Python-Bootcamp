## 1. What is a Data Structure?

A **data structure** is a way of **organising data in memory** so that the operations you need are fast.

Real-life analogies:

| Real life | Data structure | Why that shape? |
|---|---|---|
| A row of numbered lockers | **Array / list** | Jump straight to locker #57 |
| A phone contact list (search by name) | **Hash table / dict** | Find "Alice" instantly |
| A stack of plates | **Stack** | Take from the top only |
| People waiting at a bank | **Queue** | First come, first served |
| A family tree / company org chart | **Tree** | Parent → children hierarchy |
| A road map between cities | **Graph** | Anything connected to anything |
| A hospital emergency room | **Heap / priority queue** | Most urgent patient first |
| A dictionary's word index | **Trie** | Words sharing prefixes |

The **same data** can be stored in different structures - and the choice changes how fast your program is.

```python
# Same data, two structures, very different speed for "is x present?"
import time

n = 1_000_000
as_list = list(range(n))
as_set = set(as_list)
target = n - 1          # worst case for the list: last element

start = time.perf_counter()
for _ in range(100):
    target in as_list           # list: checks elements one by one -> O(n)
list_time = time.perf_counter() - start

start = time.perf_counter()
for _ in range(100):
    target in as_set            # set: hash lookup -> O(1)
set_time = time.perf_counter() - start

print(f"list: {list_time:.4f}s   set: {set_time:.6f}s")
print(f"set was ~{list_time / set_time:,.0f}x faster")
```

## 2. What is an Algorithm?

An **algorithm** is a **step-by-step procedure** to solve a problem. A cooking recipe is an algorithm.

A good algorithm is:
- **Correct** - gives the right answer for every valid input (including edge cases).
- **Efficient** - uses as little **time** and **memory** as possible.
- **Clear** - others (and future you) can understand it.

Example problem: *find a name in a phone book with 1,000,000 sorted names.*

| Algorithm | Steps in the worst case |
|---|---|
| Linear search: read every name | 1,000,000 |
| Binary search: open in the middle, discard half, repeat | **~20** |

Same data, same computer - the **algorithm** made it 50,000× faster.

```python
def linear_search_steps(sorted_names, target):
    steps = 0
    for name in sorted_names:
        steps += 1
        if name == target:
            break
    return steps

def binary_search_steps(sorted_names, target):
    lo, hi, steps = 0, len(sorted_names) - 1, 0
    while lo <= hi:
        steps += 1
        mid = (lo + hi) // 2
        if sorted_names[mid] == target:
            break
        if sorted_names[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return steps

names = sorted(f"person{i:07d}" for i in range(1_000_000))
target = names[-1]
print("linear search steps:", linear_search_steps(names, target))
print("binary search steps:", binary_search_steps(names, target))
```

## 3. Data Structures + Algorithms = Programs

> *"Algorithms + Data Structures = Programs"* - Niklaus Wirth (creator of Pascal), 1976

They always go together:
- **Binary search** only works on a **sorted array**.
- **BFS** needs a **queue**; **DFS** needs a **stack** (or recursion).
- **Dijkstra's shortest path** needs a **graph** + a **heap**.
- An **LRU cache** needs a **hash map** + a **doubly linked list**.

Choosing the right structure usually *is* the algorithm.

## 4. Why does DSA exist? (The problem of scale)

With 10 items, every approach is fast. Real systems have **millions or billions** of items:

| Input size n | O(n²) operations | O(n log n) operations | O(log n) |
|---|---|---|---|
| 1,000 | 1,000,000 | ~10,000 | ~10 |
| 1,000,000 | 10¹² (**~hours**) | ~20,000,000 (**<1 s**) | ~20 |
| 1,000,000,000 | 10¹⁸ (**centuries**) | ~3×10¹⁰ | ~30 |

DSA exists because **hardware cannot save a bad algorithm**. A faster computer gives you maybe 10× - a better algorithm can give you 1,000,000×.

```python
import math

print(f"{'n':>15} | {'n^2':>22} | {'n log n':>15} | {'log n':>6}")
for n in [10, 1_000, 1_000_000, 1_000_000_000]:
    print(f"{n:>15,} | {n*n:>22,} | {int(n*math.log2(n)):>15,} | {math.log2(n):>6.1f}")
```

## 5. Where is DSA used in real systems?

Every piece of software you use runs on these ideas:

| System | Data structure / algorithm | What it does |
|---|---|---|
| **Google Maps, Uber, Bolt** | Graphs + **Dijkstra / A\*** + heaps | Shortest / fastest route |
| **Google Search** | Inverted index (**hash maps**), **PageRank** (graph), sorting | Find & rank pages |
| **Facebook, LinkedIn, Instagram** | **Graphs** + BFS | Friends, "people you may know", degrees of connection |
| **Netflix, YouTube, Spotify recommendations** | Graphs, **heaps (top-k)**, matrices | "Because you watched…" |
| **Databases (MySQL, PostgreSQL)** | **B-trees**, hash indexes | Find a row among billions in milliseconds |
| **Redis, Memcached, browser cache** | **Hash tables** + **LRU (linked list)** | Super-fast caching |
| **Autocomplete / spell-check** | **Tries**, edit distance (**DP**) | Suggest words as you type |
| **Git** | **Graph (DAG)**, hashing (SHA), **LCS (DP)** for `git diff` | Version history and diffs |
| **Operating systems** | **Queues**, **heaps**, linked lists | CPU scheduling, memory management |
| **Browsers** | **Stacks** (back button), **trees** (HTML DOM) | Navigation, rendering pages |
| **Compilers / Python itself** | **Stacks**, **trees** (syntax tree), hash tables | Run your code; Python `dict` is a hash table |
| **Internet routing** | Graphs + Dijkstra (OSPF protocol) | Send packets the shortest way |
| **ZIP, JPEG, MP3 compression** | **Huffman coding** (greedy + heap + tree) | Smaller files |
| **Mobile money / banking (e.g. M-Pesa, MoMo)** | Hash tables, queues, sorting, graphs (fraud rings) | Fast lookups, transaction processing, fraud detection |
| **Data science & ML** | Arrays/matrices (NumPy), **trees** (decision trees), KD-trees, graphs | Training models, nearest neighbours |
| **Undo / redo in editors** | **Two stacks** | Ctrl+Z / Ctrl+Y |

## 6. Why do companies test DSA in interviews?

Companies like **Google, Meta, Amazon, Microsoft, Apple, Netflix, Uber, Bloomberg** and most fast-growing startups use **LeetCode-style DSA interviews**. Why?

1. **Problem-solving ability** - can you break an unfamiliar problem into steps?
2. **Efficiency thinking** - do you notice when code is O(n²) and fix it?
3. **Coding fluency** - can you write correct, clean code under time pressure?
4. **Communication** - can you explain trade-offs (time vs memory)?
5. **A common standard** - it's fair to compare candidates from different backgrounds.

### What a typical coding interview looks like
- 45–60 minutes, 1–2 problems, usually **medium** difficulty.
- You talk through your idea → code it → test it → state the time & space complexity.

### Where it matters in the job (not just interviews)
- **Backend engineers:** choosing caches, indexes, queues; avoiding slow queries.
- **Data engineers / data scientists:** processing millions of rows efficiently (hashing, sorting, joins, graphs).
- **ML engineers:** vector search, trees, efficient batching.
- **Mobile / frontend:** fast lists, search, rendering trees (DOM / virtual DOM).
- **Systems / infrastructure:** scheduling, networking, storage.

## 7. Why Python for DSA interviews?
- Short, readable code → less time typing, more time thinking.
- Built-in power tools: `list`, `dict`, `set`, `collections.deque`, `Counter`, `heapq`, `bisect`.
- Accepted by virtually every company for coding interviews.
- Watch out for: recursion limit (~1000), and `list.pop(0)` being O(n) - use `deque`.
