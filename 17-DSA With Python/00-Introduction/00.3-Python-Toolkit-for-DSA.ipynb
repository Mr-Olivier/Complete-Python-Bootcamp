## 1.2 Strings
Strings are **immutable** - build results in a list and `"".join(...)` at the end (O(n) instead of O(n²)).

```python
s = "Hello, World"

print(s.lower(), s.upper())
print(s.split(", "))                 # ['Hello', 'World']
print("-".join(["a", "b", "c"]))     # 'a-b-c'
print("a1!".isalnum(), "abc".isalpha(), "123".isdigit())

# char <-> number (used for 26-letter count arrays)
print(ord('a'), chr(97), ord('c') - ord('a'))   # 97 'a' 2

# build strings efficiently
parts = []
for ch in "python":
    parts.append(ch.upper())
print("".join(parts))
```

## 1.3 Dictionaries & Sets - the #1 interview tool
Average **O(1)** insert / lookup / delete. Whenever you think *"have I seen this before?"* or *"how many times?"* → hash map / set.

```python
# frequency count by hand
count = {}
for ch in "banana":
    count[ch] = count.get(ch, 0) + 1
print(count)                      # {'b': 1, 'a': 3, 'n': 2}

for key, value in count.items():
    print(key, value, end=" | ")
print()

# sets: membership & dedup
seen = set()
seen.add(3); seen.add(5)
print(3 in seen, 4 in seen)        # True False
print(set([1, 2, 2, 3]))           # {1, 2, 3}
print({1, 2, 3} & {2, 3, 4}, {1, 2} | {3})

# tuples can be dict keys / set items (lists can't!)
visited = {(0, 0), (0, 1)}         # grid cells
print((0, 1) in visited)
```

## 1.4 `collections` - Counter, defaultdict, deque
Not covered in the course but used **constantly** on LeetCode.

```python
from collections import Counter, defaultdict, deque

# Counter: frequency map in one line
c = Counter("mississippi")
print(c)                     # Counter({'i': 4, 's': 4, 'p': 2, 'm': 1})
print(c['s'], c['z'])        # 4 0  (missing key -> 0, no KeyError)
print(c.most_common(2))      # top-2 frequent
print(Counter("listen") == Counter("silent"))   # anagram check

# defaultdict: auto-creates default values
graph = defaultdict(list)
for u, v in [(1, 2), (1, 3), (2, 4)]:
    graph[u].append(v)
print(dict(graph))           # {1: [2, 3], 2: [4]}

# deque: O(1) at BOTH ends -> queue for BFS, sliding windows
q = deque([1, 2, 3])
q.append(4)        # right
q.appendleft(0)    # left
print(q.popleft(), q.pop(), q)   # 0 4 deque([1, 2, 3])
```

## 1.5 `heapq` - priority queue (min-heap)
- `heappush` / `heappop` → **O(log n)**, `h[0]` = smallest → O(1).
- Python only has a **min-heap**. For a max-heap, push **negative** values.
- Push tuples `(priority, item)` to order by priority.
- Use for: *k largest / smallest*, *top-k frequent*, *merge k sorted*, Dijkstra.

```python
import heapq

h = []
for x in [5, 1, 8, 3]:
    heapq.heappush(h, x)
print(h[0])                     # smallest = 1
print(heapq.heappop(h))         # 1

nums = [5, 1, 8, 3, 9, 2]
heapq.heapify(nums)             # list -> heap in O(n)
print(heapq.nsmallest(3, nums), heapq.nlargest(2, nums))

# max-heap trick
max_h = []
for x in [5, 1, 8]:
    heapq.heappush(max_h, -x)
print(-heapq.heappop(max_h))    # 8

# tuples: ordered by first element
tasks = []
heapq.heappush(tasks, (2, "write code"))
heapq.heappush(tasks, (1, "drink coffee"))
print(heapq.heappop(tasks))     # (1, 'drink coffee')
```

## 1.6 `bisect` - binary search on a sorted list
- `bisect_left(a, x)` → first index where `a[i] >= x`
- `bisect_right(a, x)` → first index where `a[i] > x`

```python
import bisect

a = [1, 3, 3, 5, 7]
print(bisect.bisect_left(a, 3))    # 1
print(bisect.bisect_right(a, 3))   # 3
print(bisect.bisect_left(a, 4))    # 3 -> where 4 would be inserted
bisect.insort(a, 4)                # insert keeping it sorted
print(a)
```

## 1.7 Sorting with `key` and `lambda`
Recap of `4-Functions/4.3-Lambda.ipynb` - custom sorting appears in interval, scheduling and "top k" problems.

```python
intervals = [[5, 6], [1, 3], [2, 4]]
intervals.sort(key=lambda x: x[0])         # by start
print(intervals)

words = ["banana", "kiwi", "apple", "fig"]
print(sorted(words, key=len))              # by length
print(sorted(words, key=lambda w: (len(w), w)))   # length, then alphabetical
print(sorted(words, key=lambda w: -len(w)))       # longest first

people = [("ann", 30), ("bob", 25)]
print(max(people, key=lambda p: p[1]))     # oldest
```

## 1.8 Recursion & Memoization
Recursion = function calls itself on a **smaller** input until a **base case**.
It's the foundation of **trees, DFS, backtracking and DP**.

`@cache` (from `functools`) remembers results → turns exponential recursion into linear.

```python
from functools import cache
import sys
sys.setrecursionlimit(10_000)   # Python's default (~1000) can be too low for deep recursion

def fib_slow(n):                # O(2^n)  - recomputes the same values
    if n <= 1:
        return n
    return fib_slow(n - 1) + fib_slow(n - 2)

@cache
def fib_fast(n):                # O(n)    - each n computed once
    if n <= 1:
        return n
    return fib_fast(n - 1) + fib_fast(n - 2)

print(fib_slow(20), fib_fast(90))
```

## 1.9 Classes you'll see on LeetCode
LeetCode gives you these node classes - you just need to read and use them (see `8-Class And Objects/8.1-oops.ipynb`).
Below are helpers to **build** them from lists so we can test locally.

```python
from typing import List, Optional

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

# ---------- helpers for testing (not needed on LeetCode) ----------
def build_list(values):
    dummy = ListNode()
    cur = dummy
    for v in values:
        cur.next = ListNode(v)
        cur = cur.next
    return dummy.next

def list_to_py(head):
    out = []
    while head:
        out.append(head.val)
        head = head.next
    return out

def build_tree(values):
    # level-order list like LeetCode: [3,9,20,None,None,15,7]
    if not values or values[0] is None:
        return None
    from collections import deque
    root = TreeNode(values[0])
    queue, i = deque([root]), 1
    while queue and i < len(values):
        node = queue.popleft()
        if i < len(values) and values[i] is not None:
            node.left = TreeNode(values[i]); queue.append(node.left)
        i += 1
        if i < len(values) and values[i] is not None:
            node.right = TreeNode(values[i]); queue.append(node.right)
        i += 1
    return root

def tree_to_list(root):
    from collections import deque
    out, q = [], deque([root])
    while q:
        node = q.popleft()
        if node:
            out.append(node.val); q.append(node.left); q.append(node.right)
        else:
            out.append(None)
    while out and out[-1] is None:
        out.pop()
    return out

print(list_to_py(build_list([1, 2, 3])))
print(tree_to_list(build_tree([3, 9, 20, None, None, 15, 7])))
```

## 1.10 Handy built-ins & one-liners
Small things that make interview code shorter and cleaner.

```python
# swap & multiple assignment (no temp variable)
a, b = 1, 2
a, b = b, a
print(a, b)

# unpacking
first, *middle, last = [1, 2, 3, 4, 5]
print(first, middle, last)
x, y = (3, 4)

# divmod -> quotient and remainder (digits, carries, time conversion)
print(divmod(17, 5))                 # (3, 2)
minutes, seconds = divmod(135, 60)
print(f"{minutes}m {seconds}s")

# integer division with negatives: // floors, int(a / b) truncates toward zero
print(-7 // 2, int(-7 / 2))          # -4 -3   (LeetCode usually wants truncation)

# infinity as a starting value for min/max searches
best = float("inf")
for v in [5, 3, 8]:
    best = min(best, v)
print(best)

# any / all with generators
nums = [2, 4, 6, 7]
print(any(n % 2 for n in nums), all(n > 0 for n in nums))

# zip and enumerate together; enumerate can start at 1
for i, (name, score) in enumerate(zip(["amina", "eric"], [90, 85]), start=1):
    print(i, name, score)

# max / min with key; sort a dict by its values
scores = {"amina": 90, "eric": 85, "grace": 95}
print(max(scores, key=scores.get))
print(sorted(scores.items(), key=lambda kv: kv[1], reverse=True))

# sum of a generator, counting with sum
print(sum(x * x for x in range(5)), sum(1 for c in "banana" if c == "a"))

# reversed / range backwards
print(list(reversed([1, 2, 3])), list(range(5, 0, -1)))

# string helpers
print("abc".isalpha(), "123".isdigit(), "a1".isalnum(), "Hello".count("l"), "  hi  ".strip())

# integer <-> binary / digits
print(bin(10), int("1010", 2), [int(d) for d in str(4096)])

# chained comparison and ternary
r, c, rows, cols = 2, 3, 5, 5
inside = 0 <= r < rows and 0 <= c < cols
print("inside grid" if inside else "outside")
```
